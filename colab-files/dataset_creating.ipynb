```python
import PyPDF2
import csv
import asyncio
from transformers import pipeline, T5ForConditionalGeneration, T5Tokenizer
import re

# Step 1: Extracting text from PDF
def extract_text_from_pdf(pdf_path):
    """
    Extract text from a PDF file and return it as a string.

    :param pdf_path: Path to the PDF file.
    :return: Extracted text from the PDF.
    """
    with open(pdf_path, 'rb') as file:
        reader = PyPDF2.PdfReader(file)
        text = ""
        for page in range(len(reader.pages)):
            text += reader.pages[page].extract_text()
    return text

# Step 2: Summarizing the text with adjusted max_length
summarizer = pipeline("summarization", model="facebook/bart-large-cnn", tokenizer="facebook/bart-large-cnn")

def summarize_text(text, max_length=20, min_length=10):
    """
    Summarize the text using the Hugging Face summarization model.

    :param text: The text to be summarized.
    :param max_length: The maximum length of the summary.
    :param min_length: The minimum length of the summary.
    :return: The summarized text.
    """
    summary = summarizer(text, max_length=max_length, min_length=min_length, do_sample=False)
    return summary[0]['summary_text']

# Step 3: Generate questions based on the summarized text, with a structured format
tokenizer = T5Tokenizer.from_pretrained('t5-small')
model = T5ForConditionalGeneration.from_pretrained('t5-small')

def generate_questions_from_summary(summary_text):
    """
    Generate questions based on the summarized text using the T5 model in a structured format.

    :param summary_text: The summarized text.
    :return: Generated questions as a string with structured format.
    """
    input_text = f"""
    You are a quiz generator. You are given a summary of some content. Based on the summary, generate a quiz question with four answer options and indicate the correct answer. The question should be related to the content of the summary and the options should be plausible. Please follow the exact output format below:

    Question: What is the main topic discussed in the summary?
    Options:
    1. [Option 1]
    2. [Option 2]
    3. [Option 3]
    4. [Option 4]
    Correct Answer: [Option Number]

    Here is the summary: {summary_text}
    """

    input_ids = tokenizer.encode(input_text, return_tensors='pt')
    output = model.generate(input_ids, max_length=200)
    question_output = tokenizer.decode(output[0], skip_special_tokens=True)

    return question_output

# Step 4: Parse the structured output to extract question, options, and correct answer
def parse_quiz_output(quiz_output):
    print("Generated Quiz Output:", quiz_output)

    """
    Parse the generated quiz output into a structured format (question, options, correct answer).

    :param quiz_output: The output generated by the model.
    :return: Tuple of (question, options, correct_answer)
    """
    # Check if the output is still a placeholder
    if "[Your Question Here]" in quiz_output:
        print(f"Generated output is a placeholder: {quiz_output}")
        return None, None, None

    # Define the patterns for matching each part of the quiz
    question_pattern = re.compile(r"Question: (.+?)\n")
    options_pattern = re.compile(r"Options:\n1. (.+?)\n2. (.+?)\n3. (.+?)\n4. (.+?)\n")
    correct_answer_pattern = re.compile(r"Correct Answer: (\d)")

    question_match = re.search(question_pattern, quiz_output)
    options_match = re.findall(options_pattern, quiz_output)
    correct_answer_match = re.search(correct_answer_pattern, quiz_output)

    if not question_match or not options_match or not correct_answer_match:
        print(f"Error: Unable to parse quiz output. Raw output:\n{quiz_output}")
        return None, None, None

    question = question_match.group(1).strip()
    options = options_match[0]  # We expect only one set of options
    correct_answer = correct_answer_match.group(1).strip()

    return question, options, correct_answer

# Step 5: Save the generated questions to a CSV file
def save_to_csv_incrementally(questions_and_options, filename="generated_questions.csv"):
    try:
        with open(filename, 'a', newline='') as file:
            writer = csv.writer(file)
            if file.tell() == 0:  # If file is empty, write header
                writer.writerow(["Question", "Option 1", "Option 2", "Option 3", "Option 4", "Correct Answer"])
            # Write the data
            for question, options, correct_answer in questions_and_options:
                writer.writerow([question, *options, correct_answer])
    except Exception as e:
        print(f"Error saving to CSV: {str(e)}")

# Step 6: Print questions in quiz format
def print_quiz_format(questions_and_options):
    """
    Print the questions and options in a quiz format.

    :param questions_and_options: List of questions and options.
    """
    for idx, (question, options, correct_answer) in enumerate(questions_and_options):
        print(f"Q{idx+1}: {question}")
        for i, option in enumerate(options, 1):
            print(f"  Option {i}: {option}")
        print(f"  Correct Answer: {correct_answer}")
        print()

# Step 7: Process PDF paragraphs, summarize, generate questions, and save/print results
async def process_and_generate(pdf_paragraphs, output_csv='generated_questions.csv', save_to_csv=True, print_quiz=False):
    """
    Process each paragraph from the PDF, summarize it, generate questions, and save/print results.
    """
    questions_and_answers = []

    for paragraph in pdf_paragraphs:
        # Step 1: Summarize the paragraph
        summary = summarize_text(paragraph)

        # Step 2: Generate questions from the summarized text
        quiz_output = generate_questions_from_summary(summary)

        # Print the raw quiz output for debugging
        print(f"Generated Quiz Output:\n{quiz_output}")

        # Step 3: Parse the generated output into a structured format
        question, options, correct_answer = parse_quiz_output(quiz_output)

        if question:  # Proceed only if we have a valid question
            questions_and_answers.append([question.strip(), options, correct_answer])

    # Save to CSV if required
    if save_to_csv:
        save_to_csv_incrementally(questions_and_answers, output_csv)

    # Print the questions in quiz format if required
    if print_quiz:
        print_quiz_format(questions_and_answers)

# Main function to orchestrate everything
async def main(pdf_path, output_csv='generated_questions.csv', save_to_csv=True, print_quiz=False):
    # Extract text from the PDF
    pdf_text = extract_text_from_pdf(pdf_path)

    # Break the extracted text into paragraphs
    paragraphs = pdf_text.split("\n")

    # Process paragraphs and generate questions, then save or print
    await process_and_generate(paragraphs, output_csv, save_to_csv, print_quiz)

# Example usage
pdf_path = 'Sample PDF for Text Extraction and Summarization.pdf'  # Replace with the path to your PDF file
await main(pdf_path, save_to_csv=True, print_quiz=True)  # Save to CSV and print quiz
```

```python
import requests
import pandas as pd
import json
import os
import PyPDF2  # Ensure you have this installed via pip
import logging
import time  # Added for delay/throttling
from tenacity import retry, stop_after_attempt, wait_exponential
from datetime import datetime

# --- Helper Function to Get Current Timestamp ---
def current_time():
    return datetime.now().strftime('%Y-%m-%d %H:%M:%S')

# --- Logging Configuration ---
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

# --- Error Logging to File Function ---
def log_api_error(error_message):
    with open("error_file.txt", "a", encoding="utf-8") as ef:
        ef.write(f"[{current_time()}] {error_message}\n")

# --- API Endpoint and Keys ---
GEMINI_API_URL = "https://generativelanguage.googleapis.com/v1beta/models/gemini-1.5-flash:generateContent"
API_KEY_1 = 'AIzaSyAF_15kffaTrSUy0Ad_PO3MPwKFWqpyVi8'
API_KEY_2 = 'AIzaSyByf0UmiBRJvcGg_nS-nIF5QbeSJTMV1gs'
API_KEY_3 = 'AIzaSyAAs1yTCbIByyphlan4UE6MwDxM4bhvvRU'
HEADERS = {"Content-Type": "application/json"}

# --- Retry Logic with Specific Exception Handling ---
@retry(stop=stop_after_attempt(3), wait=wait_exponential(multiplier=1, min=2, max=10))
def call_api(prompt_text, api_key, max_tokens=150):
    url = f"{GEMINI_API_URL}?key={api_key}"
    payload = {
        "contents": [{
            "parts": [{"text": prompt_text}]
        }],
        "generationConfig": {
            "temperature": 0.7,
            "maxOutputTokens": max_tokens,
            "topP": 0.95,
            "topK": 40
        }
    }
    logger.info(f"Calling API with key ending in: {api_key[-4:]}")
    try:
        response = requests.post(url, headers=HEADERS, json=payload)
        logger.info(f"API Response Status: {response.status_code}")
        # Check for 429 explicitly
        if response.status_code == 429:
            error_msg = f"Received 429 Too Many Requests. Waiting before retrying..."
            logger.error(error_msg)
            log_api_error(error_msg)
            time.sleep(10)  # Throttle delay when 429 is encountered
        response.raise_for_status()
        json_response = response.json()
        return json_response
    except requests.HTTPError as http_err:
        # Log 429 or any other HTTP errors
        if response.status_code == 429:
            error_msg = f"HTTP error 429 occurred: {http_err}"
        else:
            error_msg = f"HTTP error occurred: {http_err}"
        logger.error(error_msg)
        log_api_error(error_msg)
        raise  # Let retry handle it
    except json.JSONDecodeError as json_err:
        error_msg = f"JSON decode error: {json_err}"
        logger.error(error_msg)
        log_api_error(error_msg)
        raise
    except Exception as err:
        error_msg = f"Unexpected error: {err}"
        logger.error(error_msg)
        log_api_error(error_msg)
        raise

def safe_call_api(prompt_text, api_key, max_tokens=150):
    """
    Wrap call_api with broad exception handling as a fallback.
    """
    try:
        return call_api(prompt_text, api_key, max_tokens)
    except Exception as e:
        # Already logged in call_api; just return None here.
        logger.error("Fallback broad exception handling in safe_call_api: %s", e)
        return None

# --- Global List to Store Errored Paragraphs for Retry ---
errored_paragraphs = []  # Will hold tuples of (filename, paragraph)

# --- Core Functions for the Updated Pipeline ---
def validate_and_correct_paragraph(para):
    prompt = (
        f"Evaluate the following paragraph for suitability in quiz generation. "
        f"Respond with 'true' if it's suitable; otherwise, respond with 'false' and provide a corrected version:\n\n{para}"
    )
    print(f"[{current_time()}] Validating paragraph...")
    logger.info("Validating paragraph...")
    result = safe_call_api(prompt, API_KEY_1, max_tokens=500)
    if result:
        output = result.get("candidates", [{}])[0].get("content", {}).get("parts", [{}])[0].get("text", "")
        if output.lower().startswith("true"):
            logger.info("Paragraph is valid.")
            return True, para
        elif output.lower().startswith("false"):
            corrected_para = output[5:].strip()  # Assumes corrected version starts after 'false'
            logger.info("Paragraph corrected.")
            return False, corrected_para
    logger.error("Paragraph validation failed.")
    return False, para

def generate_question_options(para):
    prompt = (
        "Generate a quiz question and four answer options based on the following paragraph. "
        "Return your answer in exactly the following format:\n\n"
        "Question: <your question here>\n"
        "Options:\n"
        "1. <option 1>\n"
        "2. <option 2>\n"
        "3. <option 3>\n"
        "4. <option 4>\n\n"
        "Paragraph:\n" + para
    )
    print(f"[{current_time()}] Generating question and options...")
    logger.info("Generating question and options...")
    result = safe_call_api(prompt, API_KEY_2, max_tokens=150)
    if result:
        output = result.get("candidates", [{}])[0].get("content", {}).get("parts", [{}])[0].get("text", "")
        if "Options:" in output:
            parts = output.split("Options:")
            question_text = parts[0].replace("Question:", "").strip()
            options_text = parts[1].strip()
            options = []
            for line in options_text.splitlines():
                line = line.strip()
                if any(line.startswith(f"{i}.") for i in range(1, 5)):
                    option = line.split(".", 1)[-1].strip()
                    options.append(option)
            return question_text, options, ""
    logger.error("Question generation failed.")
    return "", [], ""

def correct_quiz_format(para, question, options, candidate_answer):
    prompt = (
        "Correct any grammatical errors and formatting issues in the following quiz item. "
        "Ensure that the output is exactly in the following format (with no extra text):\n\n"
        "Question: <your corrected question>\n"
        "Options:\n"
        "1. <option 1>\n"
        "2. <option 2>\n"
        "3. <option 3>\n"
        "4. <option 4>\n"
        "Correct Answer: <digit>\n\n"
        "Paragraph:\n" + para + "\n\n"
        "Quiz Item (as generated):\n"
        "Question: " + question + "\n" +
        "Options:\n" +
        "1. " + options[0] + "\n" +
        "2. " + options[1] + "\n" +
        "3. " + options[2] + "\n" +
        "4. " + options[3] + "\n" +
        "Generated Answer: " + (candidate_answer if candidate_answer else "Not provided") + "\n"
    )
    print(f"[{current_time()}] Correcting quiz format...")
    logger.info("Correcting quiz format...")
    result = safe_call_api(prompt, API_KEY_3, max_tokens=150)
    if result:
        output = result.get("candidates", [{}])[0].get("content", {}).get("parts", [{}])[0].get("text", "")
        corr_question = ""
        corr_options = ["", "", "", ""]
        corr_answer = ""
        for line in output.splitlines():
            if line.startswith("Question:"):
                corr_question = line.replace("Question:", "").strip()
            elif line.startswith("1."):
                corr_options[0] = line.split(".", 1)[-1].strip()
            elif line.startswith("2."):
                corr_options[1] = line.split(".", 1)[-1].strip()
            elif line.startswith("3."):
                corr_options[2] = line.split(".", 1)[-1].strip()
            elif line.startswith("4."):
                corr_options[3] = line.split(".", 1)[-1].strip()
            elif line.startswith("Correct Answer:"):
                corr_answer = line.replace("Correct Answer:", "").strip()
        return corr_question, corr_options, corr_answer
    logger.error("Quiz format correction failed.")
    return question, options, candidate_answer

def process_paragraph(para, source_pdf):
    logger.info("Processing paragraph...")
    is_valid, corrected_para = validate_and_correct_paragraph(para)
    if not is_valid:
        print(f"[{current_time()}] Paragraph was corrected.")
    if is_valid or corrected_para != para:
        question, options, candidate_answer = generate_question_options(corrected_para)
        if question and options and len(options) == 4:
            corrected_question, corrected_options, corrected_answer = correct_quiz_format(
                corrected_para, question, options, candidate_answer
            )
            logger.info("Quiz generated successfully.")
            print(f"[{current_time()}] Quiz generated successfully for one paragraph.")
            return {
                "paragraph": corrected_para,
                "question": corrected_question,
                "options": corrected_options,
                "correct_answer": corrected_answer,
                "source_pdf": source_pdf
            }
    logger.error("Failed to generate quiz from paragraph.")
    return None

def extract_text_from_pdf(pdf_path):
    text = ""
    try:
        with open(pdf_path, "rb") as f:
            pdf_reader = PyPDF2.PdfReader(f)
            for page in pdf_reader.pages:
                page_text = page.extract_text()
                if page_text:
                    text += page_text + "\n"
    except Exception as e:
        logger.error("Failed to read PDF %s: %s", pdf_path, e)
        print(f"[{current_time()}] Error reading PDF {pdf_path}")
    return text

def format_for_finetuning(quiz_item):
    """
    Format a quiz item into a predefined schema for model fine-tuning.
    Schema:
    {
        "prompt": "Generate quiz question: <paragraph>",
        "completion": "Question: <question> Options: <option1> | <option2> | <option3> | <option4> Correct Answer: <digit>"
    }
    """
    completion_options = " | ".join(quiz_item["options"])
    return {
        "prompt": f"Generate quiz question: {quiz_item['paragraph']}",
        "completion": f"Question: {quiz_item['question']} Options: {completion_options} Correct Answer: {quiz_item['correct_answer']}"
    }

def validate_quiz_item(item):
    """
    Ensure all required fields in the quiz item conform to the schema.
    """
    required_fields = ["paragraph", "question", "options", "correct_answer"]
    if not all(field in item and item[field] for field in required_fields):
        return False
    # Additional validation: exactly 4 options and non-empty strings.
    if len(item["options"]) != 4 or not all(isinstance(opt, str) and opt.strip() for opt in item["options"]):
        return False
    return True

# --- Main Function for Updated Dataset Creation ---
def main():
    start_time = current_time()
    print(f"[{start_time}] Starting updated processing of PDF files...")
    logger.info("Starting updated processing of PDF files.")

    input_directory = "."  # Adjust if PDFs are in a subfolder
    quiz_dataset = []
    invalid_paragraphs = []
    # List to store paragraphs that errored (for later retry)
    paragraphs_to_retry = []

    # Collect all PDF files from the directory
    all_pdfs = [f for f in os.listdir(input_directory) if f.lower().endswith(".pdf")]
    total_pdfs = len(all_pdfs)
    print(f"[{current_time()}] Total PDF files found: {total_pdfs}")
    logger.info("Total PDF files found: %d", total_pdfs)

    for index, filename in enumerate(all_pdfs):
        file_path = os.path.join(input_directory, filename)
        remaining = total_pdfs - index - 1
        print(f"[{current_time()}] Processing PDF {index+1}/{total_pdfs}: {filename}")
        logger.info("Processing PDF %d/%d: %s", index+1, total_pdfs, filename)
        logger.info("PDFs left: %d", remaining)

        # Extract text from PDF and split into paragraphs
        textbook_data = extract_text_from_pdf(file_path)
        logger.info("Extracted text length: %d", len(textbook_data))

        paragraphs = [p.strip() for p in textbook_data.split("\n\n") if p.strip()]
        total_paragraphs = len(paragraphs)
        print(f"[{current_time()}] {total_paragraphs} paragraphs created from {filename}.")
        logger.info("Split into %d paragraphs", total_paragraphs)

        valid_count = 0
        invalid_count = 0

        for idx, para in enumerate(paragraphs):
            print(f"[{current_time()}] Processing Paragraph {idx+1}/{total_paragraphs} from {filename}...")
            logger.info("Processing Paragraph %d/%d...", idx+1, total_paragraphs)
            result = process_paragraph(para, filename)
            if result:
                if validate_quiz_item(result):
                    quiz_dataset.append(result)
                    valid_count += 1
                else:
                    logger.error("Quiz item did not pass validation for %s", filename)
                    invalid_paragraphs.append(f"[{filename}] {para}")
                    invalid_count += 1
            else:
                # Store the paragraph for later retry
                paragraphs_to_retry.append((filename, para))
                invalid_count += 1

            # Throttle: Wait a short period after each paragraph to avoid rapid consecutive requests
            time.sleep(1)

        print(f"[{current_time()}] Finished processing {filename}: {total_paragraphs} paragraphs processed, {valid_count} valid quizzes, {invalid_count} invalid.")
        logger.info("Finished processing %s: %d paragraphs processed, %d valid, %d invalid.", filename, total_paragraphs, valid_count, invalid_count)

    # Retry errored paragraphs after processing all PDFs.
    if paragraphs_to_retry:
        print(f"[{current_time()}] Retrying {len(paragraphs_to_retry)} errored paragraphs...")
        logger.info("Retrying %d errored paragraphs...", len(paragraphs_to_retry))
        retry_success = 0
        for source_pdf, para in paragraphs_to_retry:
            # Add an extra delay between retries to throttle requests further
            time.sleep(5)
            result = process_paragraph(para, source_pdf)
            if result and validate_quiz_item(result):
                quiz_dataset.append(result)
                retry_success += 1
            else:
                invalid_paragraphs.append(f"[{source_pdf}] {para}")
        print(f"[{current_time()}] Retried paragraphs: {retry_success} succeeded.")
        logger.info("Retried paragraphs: %d succeeded.", retry_success)

    if quiz_dataset:
        df = pd.DataFrame(quiz_dataset)
        df.to_csv("quiz_dataset.csv", index=False)
        with open("quiz_dataset.json", "w", encoding="utf-8") as jf:
            json.dump(quiz_dataset, jf, indent=4)

        finetune_dataset = [format_for_finetuning(item) for item in quiz_dataset]
        with open("finetune_dataset_2.jsonl", "w", encoding="utf-8") as f:
            for record in finetune_dataset:
                f.write(json.dumps(record) + "\n")
        print(f"[{current_time()}] Quiz dataset saved with {len(quiz_dataset)} entries.")
        logger.info("Quiz dataset saved with %d entries", len(quiz_dataset))
    else:
        print(f"[{current_time()}] No valid quiz data generated.")
        logger.warning("No valid quiz data generated.")

    if invalid_paragraphs:
        with open("invalid_paragraphs.txt", "w", encoding="utf-8") as f:
            for para in invalid_paragraphs:
                f.write(para + "\n\n")
        print(f"[{current_time()}] Saved {len(invalid_paragraphs)} invalid paragraphs to invalid_paragraphs.txt")
        logger.warning("Saved %d invalid paragraphs to invalid_paragraphs.txt", len(invalid_paragraphs))

    end_time = current_time()
    print(f"[{end_time}] Finished processing of PDF files.")
    logger.info("Finished processing of PDF files.")

if __name__ == "__main__":
    main()
```
